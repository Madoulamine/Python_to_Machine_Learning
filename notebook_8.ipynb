# Jour 9 - 23 Juillet 2026 : Visualisation avec Matplotlib

Un graphique vaut mille nombres ! **Matplotlib** est la bibliothèque de base pour dessiner des courbes et des graphiques en Python.

## 1. Courbe simple (Line plot)
Idéal pour montrer l'évolution d'une donnée dans le temps.

```python
import matplotlib.pyplot as plt

annees = [2018, 2019, 2020, 2021, 2022, 2023]
ventes = [150, 180, 160, 210, 250, 280]

plt.plot(annees, ventes, color='blue', marker='o')
plt.title("Évolution des ventes")
plt.xlabel("Années")
plt.ylabel("Nombre de ventes")
plt.show() # Affiche le graphique
```

## 2. Diagramme en barres
Parfait pour comparer des catégories distinctes.

```python
produits = ['Pommes', 'Bananes', 'Cerises']
quantites = [45, 30, 15]

plt.bar(produits, quantites, color=['red', 'yellow', 'pink'])
plt.title("Stock de fruits")
plt.show()
```

## 🏆 Défi du Jour 9

1. Créez deux listes : `jours` (Lundi, Mardi, Mercredi, Jeudi) et `temperature` (20, 22, 19, 24).
2. Tracez une courbe (`plt.plot`) reliant les températures en fonction des jours.
3. Ajoutez un titre "Météo de la semaine" et les étiquettes pour les axes X et Y.
4. Affichez le graphique.

```python
# Écrivez votre code pour le défi ici
import matplotlib.pyplot as plt

# 1. Créer les listes
jours = ["Lundi", "Mardi", "Mercredi", "Jeudi"]
temperature = [20, 22, 19, 24]

# 2. Tracer la courbe
plt.plot(jours, temperature)

# 3. Ajouter un titre et les étiquettes des axes
plt.title("Météo de la semaine")
plt.xlabel("Jours")
plt.ylabel("Température (°C)")

# 4. Afficher le graphique
plt.show()
```
